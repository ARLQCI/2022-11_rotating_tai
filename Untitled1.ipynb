```python
import sympy
```

```python
from sympy import sin
```

```python
w0, t_r, t = sympy.symbols('omega_0, t_r, t', is_real=True)
π = sympy.pi
```

```python
α = sympy.symbols('alpha', is_real=True, positive=True)
```

```python
ϕ = w0 * sin(α * t)**2
```

```python
ϕ
```

```python
sol = sympy.integrate(ϕ, t)
```

```python
sol.simplify().expand().subs({α: π / (2*t_r)})
```

```python
sol.simplify().expand().subs({α: π / (2*t_r), w0: 2*π})
```
